```python
from pathlib import Path

import pandas as pd
import xarray as xr
from IPython.display import display

DATA_DIR = Path(r"E:\glacier_research\datasets\era5_land_nepal_climate_1990-2026")
OUTPUT_DIR = Path(r"E:\glacier_research\datasets_to_csv\era5_land_nepal_climate_1990-2026_csv")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
FILES = {
    "data_stream-moda": DATA_DIR / "data_stream-moda.nc",
    "geo": DATA_DIR / "geo.area-subset.39.3.105.15.9.60.8.nc",
    "cicecap": DATA_DIR / "cicecap.area-subset.39.3.105.15.9.60.8.nc",
}

# One row per timestamp/grid-cell, with coordinates and variables as columns.
dataframes = {}
for name, path in FILES.items():
    with xr.open_dataset(path) as dataset:
        dataframes[name] = dataset.to_dataframe().reset_index()

pd.set_option("display.max_columns", None)

for name, dataframe in dataframes.items():
    print(f"{name}: {dataframe.shape[0]:,} rows x {dataframe.shape[1]} columns")
    display(dataframe)
```

```python
# Save all three DataFrames as CSV files
for name, dataframe in dataframes.items():
    output_file = OUTPUT_DIR / f"{name}.csv"
    dataframe.to_csv(output_file, index=False)
    print(f"Saved: {output_file}")
```
