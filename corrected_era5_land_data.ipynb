```python
import xarray as xr
import os
import pandas as pd

# Three folders with nc files
folders = [
    r"E:\glacier_research\datasets\era5_land_nepal_climate_1990-2026\corrected_data_2022_SeptoDec",
    r"E:\glacier_research\datasets\era5_land_nepal_climate_1990-2026\corrected_data_2023_JantoDec", 
    r"E:\glacier_research\datasets\era5_land_nepal_climate_1990-2026\corrected_data_2024_JantoFeb"
]

# Open and combine vertically
datasets = []
for folder in folders:
    nc_path = os.path.join(folder, "data_stream-mnth.nc")
    ds = xr.open_dataset(nc_path)
    datasets.append(ds)

combined = xr.concat(datasets, dim="valid_time")
print("Combined dimensions:", combined.sizes)

# Convert to dataframe and save as CSV
df = combined.to_dataframe().reset_index()
output_path = r"E:\glacier_research\datasets_to_csv\corrected_era5_land_data\combined_corrected_data.csv"
df.to_csv(output_path, index=False)
print(f"CSV saved to: {output_path}")
print(f"Shape: {df.shape}")
```
